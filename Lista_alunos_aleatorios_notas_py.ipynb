```python
#### Exercício 3

#Você recebeu um dicionário com os nomes dos alunos e suas respectivas notas.
# Escreva uma função que calcula a média da turma e retorna uma lista com os nomes dos alunos que tiveram nota acima da média.

#!pip install Faker
from faker import Faker
import random
import pandas as pd

quant_alunos = 10
def dict_fiticios(quant_alunos):
    # convertendo apenas para nomes brasileiros
    fake = Faker('pt_BR')
    lista_nome_de_alunos = []
    dicionario_alunos = {}
    # adicionando os nomes aleatorios em uma lista
    for c in range(quant_alunos):
        lista_nome_de_alunos.append(fake.name())
    contador = 0

    #adicionando as nota aleatorias na lista
    for nota in range(1,quant_alunos+1):
        nota_aleatoria = random.uniform(3,10)
        nota_aleatoria = round(nota_aleatoria,1)
        lista_nome_de_alunos.insert(contador+ nota,nota_aleatoria)
        contador +=1


    #transformando a lista em dicionario
    for i in range(0,len(lista_nome_de_alunos),2):
        dicionario_alunos[lista_nome_de_alunos[i]] = lista_nome_de_alunos[i+1]

    return dicionario_alunos

base_alunos = dict_fiticios(quant_alunos)

# criando um DataFrame e inserindo as colunas
base = pd.DataFrame(list(base_alunos.items()),columns=['Aluno','Nota'])

# filtro alunos aprovados com nota maior que media da turma ordenado por nota maior
Alunos_Aprovados = base[base['Nota']>=base.Nota.mean()].sort_values(by='Nota',ascending=False)
Alunos_Aprovados.head()
```
