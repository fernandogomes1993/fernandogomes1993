```python
class Carrinho:
  def __init__(self):
    self._produtos = []

  def adicionar_produtos(self,*produto):
    self._produtos += produto

  def lista_produtos(self):
    for quat,listas_com_produto in enumerate(self._produtos):
      print(f'{quat+1} - ',listas_com_produto.nome,listas_com_produto.preco)

  def preco_total (self):
    return round(sum([preco_produtos.preco for preco_produtos in self._produtos]),2)

class Produto:
  def __init__(self, nome, preco):
    self.nome = nome
    self.preco = preco


carrinho = Carrinho()

# monte o carrinho
arroz = Produto('arroz',39.9)
feijao = Produto('feijao',19.9)
milho = Produto('Milho',29.9)

# adicione ao carrinho
carrinho.adicionar_produtos(arroz,feijao,milho)

# mostrndo o carrinho
print('--- lista de produtos ---')
carrinho.lista_produtos()
print()
print(f'Foram adicionado {len(carrinho._produtos)} produtos ao carrinho total da comprar R$ {carrinho.preco_total()}')
```
